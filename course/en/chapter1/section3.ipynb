# Transformers, what can they do?

Install the Transformers, Datasets, and Evaluate libraries to run this notebook.

```python
from transformers import pipeline

classifier = pipeline("sentiment-analysis")
classifier("I hate HuggingFace")
```

```python
classifier(
    ["I love you", "I don't love you!"]
)
```

```python
from transformers import pipeline

classifier = pipeline("zero-shot-classification")
classifier(
    "I support Pakatan Harapan",
    candidate_labels=["education", "politics", "business"],
)
```

```python
from transformers import pipeline

generator = pipeline("text-generation",device=0)
generator("This morning, I would like to", max_new_tokens=50)
```

```python
from transformers import pipeline

generator = pipeline("text-generation", model="distilgpt2")
generator(
    "Every morning, I would",
    max_length=30,
    num_return_sequences=2,
)
```

```python
from transformers import pipeline

unmasker = pipeline("fill-mask")
unmasker("Best singers in the world like <mask> is amazing", top_k=2)
```

```python
# Step 1: fill description
step1 = unmasker("A singer who is <mask>.", top_k=1)
description = step1[0]["token_str"].strip()

# Step 2: use that description as context for name
step2 = unmasker(f"A singer who is {description}, like <mask>.", top_k=2)

for r in step2:
    print(r["sequence"])
```

```python
from transformers import pipeline

ner = pipeline("ner", aggregation_strategy="simple")
ner("I am Quinton and I like to eat Mcdonald's.") # BERT was never trained to see "McDonald's" as one clean token — the apostrophe breaks it apart and confuses the entity labeling.
```
